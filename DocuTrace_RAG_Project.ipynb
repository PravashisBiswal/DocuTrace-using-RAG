```python
# laoding the document
loader = UnstructuredFileLoader("python_inbuildfunction.pdf")
```

```python
text_splitter = CharacterTextSplitter(
    chunk_size=2000,
    chunk_overlap=400
)
```

```python
document = loader.load()
texts = text_splitter.split_documents(document)
```

```python
embeddings = HuggingFaceEmbeddings()
```

```python
persist_directory = "vector_db"
```

```python
vectordb = Chroma.from_documents(
    documents=texts,
    embedding=embeddings,
    persist_directory=persist_directory
)
```

```python
# retriever
retriever = vectordb.as_retriever()
```

```python
# llm from groq
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)
```

```python
# invoke the qa chain and get a response for user query
query = "what is inheritance  from this pdf "
response = qa_chain.invoke({"query": query})
```

```python
import gradio as gr

# Chat Function
def chatbot_response(question):

    try:
        response = qa_chain.invoke({"query": question})

        answer = response["result"]

        sources = []

        for doc in response["source_documents"]:
            if "source" in doc.metadata:
                sources.append(doc.metadata["source"])

        source_text = "\n".join(list(set(sources)))

        final_response = f"""
{answer}

-------------------
Source:
{source_text}
"""

        return final_response

    except Exception as e:
        return f"Error: {str(e)}"


# Gradio UI
demo = gr.Interface(
    fn=chatbot_response,
    inputs=gr.Textbox(
        lines=2,
        placeholder="Ask questions from your PDF..."
    ),
    outputs="text",
    title="DocuTrace",
    description="LLMOps Project using LangChain + Groq + ChromaDB + Gradio",
    theme="soft"
)

demo.launch(debug=True)
```
